```python
import os
import numpy as np
import librosa
import matplotlib.pyplot as plt
from collections import defaultdict, Counter
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
RAVDESS_DIR = "RAVDESS"
CREMAD_DIR = "CREMA-D"
EMOTIONS = ["happy", "sad"]
USE_FAKE_DATA = False
```

```python
# A1(a) - ENCODING :- it will turn one audio file to a list of numbers

def get_mfcc(file_path):
    signal, sr = librosa.load(file_path, sr=None)
    mfcc = librosa.feature.mfcc(y=signal, sr=sr, n_mfcc=13)
    return np.mean(mfcc.T, axis=0)
```

```python
# RAVDESS - first dataset
def ravdess_label(filename):
    codes = {"01": "neutral", "02": "calm", "03": "happy", "04": "sad",
             "05": "angry", "06": "fearful", "07": "disgust", "08": "surprised"}
    return codes[filename.split("-")[2]]
```

```python
# CREMA-D - second dataset
def cremad_label(filename):
    codes = {"ANG": "angry", "DIS": "disgust", "FEA": "fearful",
             "HAP": "happy", "NEU": "neutral", "SAD": "sad"}
    return codes[filename.split("_")[2]]
```

```python
def load_ravdess(folder, wanted_emotions):
    X, y = [], []

    for root, _, files in os.walk(folder):
        for f in files:

            if not f.endswith(".wav"):
                continue

            label = ravdess_label(f)

            if label in wanted_emotions:
                X.append(
                    get_mfcc(os.path.join(root, f))
                )
                y.append(label)

    return X, y
```

```python
def load_cremad(folder, wanted_emotions):
    X, y = [], []

    for root, _, files in os.walk(folder):
        for f in files:

            if not f.endswith(".wav"):
                continue

            label = cremad_label(f)

            if label in wanted_emotions:
                X.append(
                    get_mfcc(os.path.join(root, f))
                )
                y.append(label)

    return X, y
```

```python
def fake_dataset(n=200, features=13):
    rng = np.random.default_rng(0)

    class1 = rng.normal(
        0, 1, (n // 2, features)
    )

    class2 = rng.normal(
        3, 1, (n // 2, features)
    )

    X = np.vstack([class1, class2])

    y = ["happy"] * (n // 2) + ["sad"] * (n // 2)

    return X, y
```

```python
# A1a continued - turns "happy"/"sad" text labels into 0/1 numbers
def encode_labels(y):
    classes = sorted(set(y))
    mapping = {c: i for i, c in enumerate(classes)}

    return np.array([
        mapping[val] for val in y
    ]), mapping
```

```python
# A1b - DATA IMPUTATION :- it fills the missing values using mean, median, or mode

def fill_missing(X, method="mean"):
    X = np.array(X, dtype=float)
    out = X.copy()

    for col in range(X.shape[1]):

        values = X[:, col]
        missing = np.isnan(values)

        if not missing.any():
            continue

        if method == "mean":
            fill = np.nanmean(values)

        elif method == "median":
            fill = np.nanmedian(values)

        else:
            fill = Counter(
                values[~missing]
            ).most_common(1)[0][0]

        out[missing, col] = fill

    return out
```

```python
# A1c - DISTANCE CALCULATION :- it uses three different distance metrics and picks any at prediction time

def euclidean(a, b):
    return np.sqrt(
        np.sum((a - b) ** 2)
    )
```

```python
def manhattan(a, b):
    return np.sum(
        np.abs(a - b)
    )
```

```python
def minkowski(a, b, p=3):
    return np.sum(
        np.abs(a - b) ** p
    ) ** (1 / p)
```

```python
# distance from one test point to every training point

def all_distances(X_train, point, metric="euclidean"):
    dists = np.zeros(len(X_train))

    for i, row in enumerate(X_train):

        if metric == "euclidean":
            dists[i] = euclidean(row, point)

        elif metric == "manhattan":
            dists[i] = manhattan(row, point)

        else:
            dists[i] = minkowski(row, point)

    return dists
```

```python
# A1d - SORTING :- we are using 3 DSA sorting algorithms

def bubble_sort(pairs):
    arr = pairs[:]

    for i in range(len(arr)):

        for j in range(len(arr) - i - 1):

            if arr[j][0] > arr[j + 1][0]:
                arr[j], arr[j + 1] = (
                    arr[j + 1],
                    arr[j]
                )

    return arr
```

```python
def insertion_sort(pairs):
    arr = pairs[:]

    for i in range(1, len(arr)):

        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j][0] > key[0]:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr
```

```python
def merge_sort(pairs):

    if len(pairs) <= 1:
        return pairs

    mid = len(pairs) // 2

    left = merge_sort(pairs[:mid])
    right = merge_sort(pairs[mid:])

    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):

        if left[i][0] <= right[j][0]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    return result + left[i:] + right[j:]
```

```python

# config parameter picks which sorting algorithm runs
def sort_distances(dists, algo="merge"):

    pairs = list(
        zip(dists, range(len(dists)))
    )

    if algo == "bubble":
        return bubble_sort(pairs)

    elif algo == "insertion":
        return insertion_sort(pairs)

    else:
        return merge_sort(pairs)
```

```python
# A1e - IDENTIFY NEIGHBORS :- it will grab the closest k points and breaks ties by earliest index

def get_neighbors(sorted_pairs, k):

    cutoff = sorted_pairs[k - 1][0]

    sure = [
        p for p in sorted_pairs
        if p[0] < cutoff
    ]

    tied = sorted(
        [
            p for p in sorted_pairs
            if p[0] == cutoff
        ],
        key=lambda p: p[1]
    )

    return sure + tied[:k - len(sure)]
```

```python
# A1f - CLASS EVALUATION AND ASSIGNMENT :- majority voting + tie-break

def majority_vote(neighbors, y_train):

    counts = defaultdict(int)

    for dist, idx in neighbors:
        counts[y_train[idx]] += 1

    top = max(counts.values())

    winners = [
        c for c, v in counts.items()
        if v == top
    ]

    if len(winners) == 1:
        return winners[0]


    for dist, idx in sorted(neighbors):

        if y_train[idx] in winners:
            return y_train[idx]
```

```python
# A2 - WEIGHTED kNN :- the closer neighbors will get more weight

def weighted_vote(neighbors, y_train):

    weights = defaultdict(float)

    for dist, idx in neighbors:

        weights[y_train[idx]] += (
            1 / (dist + 1e-6)
        )

    top = max(weights.values())

    winners = [
        c for c, v in weights.items()
        if v == top
    ]

    if len(winners) == 1:
        return winners[0]

    # tie-break using closest neighbor
    for dist, idx in sorted(neighbors):

        if y_train[idx] in winners:
            return y_train[idx]
```

```python
# A7 - FIT / PREDICT / SCORE FUNCTIONS

def fit(X_train, y_train):
    return {
        "X": np.array(X_train, dtype=float),
        "y": np.array(y_train)
    }
```

```python
def predict(
    model,
    X_test,
    k,
    metric="euclidean",
    algo="merge",
    weighted=False
):

    preds = []

    for point in np.array(X_test, dtype=float):

        dists = all_distances(
            model["X"],
            point,
            metric
        )

        sorted_pairs = sort_distances(
            dists,
            algo
        )

        neighbors = get_neighbors(
            sorted_pairs,
            k
        )

        vote = (
            weighted_vote
            if weighted
            else majority_vote
        )

        preds.append(
            vote(
                neighbors,
                model["y"]
            )
        )

    return np.array(preds)
```

```python
def Score(y_true, y_pred):
    return np.mean(
        np.array(y_true) == np.array(y_pred)
    )
```

```python
# ---------------- main ----------------

if __name__ == "__main__":

    # load data
    if USE_FAKE_DATA:

        raw_X, raw_y = fake_dataset()

        print("using fake data")

    else:

        rav_X, rav_y = load_ravdess(
            RAVDESS_DIR,
            EMOTIONS
        )

        crema_X, crema_y = load_cremad(
            CREMAD_DIR,
            EMOTIONS
        )

        raw_X = rav_X + crema_X
        raw_y = rav_y + crema_y

        if len(raw_X) == 0:

            print("no audio files found")
            print("using fake data")

            raw_X, raw_y = fake_dataset()

        else:

            print(
                f"loaded {len(raw_X)} audio clips"
            )

    # A1a - encoding
    y_encoded, label_map = encode_labels(raw_y)

    print(
        "labels:",
        label_map
    )

    # A1b - data imputation
    X_clean = fill_missing(
        np.array(raw_X),
        method="mean"
    )

    # A3 - train test split
    X_train, X_test, y_train, y_test = train_test_split(
        X_clean,
        y_encoded,
        test_size=0.3,
        random_state=42
    )

    # A4 - train custom kNN
    model = fit(
        X_train,
        y_train
    )

    # A5 and A6 - score and predict
    my_preds = predict(
        model,
        X_test,
        k=3
    )

    print(
        "my knn accuracy:",
        Score(y_test, my_preds)
    )

    # sklearn kNN
    sk_model = KNeighborsClassifier(
        n_neighbors=3
    )

    sk_model.fit(
        X_train,
        y_train
    )

    print(
        "sklearn knn accuracy:",
        sk_model.score(
            X_test,
            y_test
        )
    )

    # A8 - compare for different k values
    k_range = range(1, 16)

    my_acc = []
    sk_acc = []

    for k in k_range:

        preds = predict(
            model,
            X_test,
            k
        )

        my_acc.append(
            Score(
                y_test,
                preds
            )
        )

        temp_model = KNeighborsClassifier(
            n_neighbors=k
        )

        temp_model.fit(
            X_train,
            y_train
        )

        sk_acc.append(
            temp_model.score(
                X_test,
                y_test
            )
        )

    # A8 plot
    plt.figure()

    plt.plot(
        k_range,
        my_acc,
        marker="o",
        label="my knn"
    )

    plt.plot(
        k_range,
        sk_acc,
        marker="s",
        label="sklearn knn"
    )

    plt.xlabel("k")
    plt.ylabel("accuracy")
    plt.title("my knn vs sklearn knn")
    plt.legend()
    plt.grid()

    plt.savefig(
        "compare_sklearn.png"
    )

    # A9 - weighted kNN
    weighted_acc = []

    for k in k_range:

        weighted_preds = predict(
            model,
            X_test,
            k,
            weighted=True
        )

        weighted_acc.append(
            Score(
                y_test,
                weighted_preds
            )
        )

    # A9 plot
    plt.figure()

    plt.plot(
        k_range,
        my_acc,
        marker="o",
        label="normal knn"
    )

    plt.plot(
        k_range,
        weighted_acc,
        marker="^",
        label="weighted knn"
    )

    plt.xlabel("k")
    plt.ylabel("accuracy")
    plt.title("normal knn vs weighted knn")
    plt.legend()
    plt.grid()

    plt.savefig(
        "compare_weighted.png"
    )

    plt.show()
```
